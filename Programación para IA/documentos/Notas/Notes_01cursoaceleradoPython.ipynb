## 1. Valores, variables y tipos

**Idea clave.** Un nombre es una *etiqueta* que apunta a un objeto; no es una
caja con un tipo fijo. El tipo vive en el objeto, no en el nombre.

**Cuándo.** Siempre que eliges un nombre de variable. Un buen nombre comunica
*intención* y evita que el significado cambie a mitad del programa.

**Tipos básicos del día a día:**

| Tipo | Ejemplo | Para qué |
|---|---|---|
| `int` | `42`, `1_000` | conteos, índices, cantidades enteras |
| `float` | `0.91`, `3.14` | scores, métricas, medidas |
| `str` | `"texto"` | texto, etiquetas, rutas |
| `bool` | `True`, `False` | banderas, condiciones |
| `NoneType` | `None` | "todavía no hay valor" / ausencia |

🧠 **Para IA/datos:** `None` representa *"este dato no se midió"*. No lo
sustituyas por `0`, `-1` o `""`: eso introduce sesgos silenciosos.

```python
modelo = "clasificador de reseñas"   # str
precision = 0.91                     # float
ejemplos = 1_250                     # int (el _ es separador legible)
esta_entrenado = True                # bool
sin_descripcion = None               # NoneType

for nombre, valor in {
    "modelo": modelo, "precision": precision, "ejemplos": ejemplos,
    "esta_entrenado": esta_entrenado, "sin_descripcion": sin_descripcion,
}.items():
    print(f"{nombre:16} -> {valor!r:28} ({type(valor).__name__})")
```

```python
# Un mismo nombre PUEDE reapuntar a otro tipo (Python lo permite)...
dato = 10
print(dato, type(dato).__name__)
dato = "ahora texto"          # legal, pero cambiar el significado confunde
print(dato, type(dato).__name__)
# 🧠 Legal != recomendable: mantén un nombre con un solo significado.
```

## 2. Operadores e identidad (`is` vs `==`)

**Idea clave.** `==` compara **valores**; `is` compara **identidad** (si son el
*mismo* objeto en memoria). Usa `is` solo con `None` (y singletons).

| Quiero preguntar… | Operador |
|---|---|
| ¿dos textos/números son iguales? | `==` |
| ¿falta el valor? | `x is None` |
| ¿es exactamente el mismo objeto (alias)? | `is` |

⚠️ Los `float` tienen error de redondeo: compara con `math.isclose`.

```python
# Aritméticos menos obvios
print("división entera 7 // 2 =", 7 // 2)   # 3
print("residuo       7 %  2 =", 7 % 2)       # 1  (par/impar, "cada N")
print("potencia      2 ** 10 =", 2 ** 10)    # 1024

# Comparación encadenada (muy pitónico)
edad = 25
print("0 <= edad < 120 ->", 0 <= edad < 120)

# Identidad: SOLO para None
valor = None
print("valor is None  ->", valor is None)   # ✅
```

```python
import math
print("0.1 + 0.2 == 0.3        ->", 0.1 + 0.2 == 0.3)          # False ⚠️
print("suma real               ->", 0.1 + 0.2)                  # 0.3000...4
print("math.isclose(...)       ->", math.isclose(0.1 + 0.2, 0.3))  # True ✅
```

## 3. Cadenas y f-strings

**Idea clave.** Las cadenas son **inmutables** (los métodos devuelven copias).
Las f-strings arman texto legible insertando valores entre llaves.

**Mini-lenguaje de formato dentro de `{}`:**

| Formato | Entrada | Salida |
|---|---|---|
| `{x:.2f}` | `3.14159` | `3.14` |
| `{x:.1%}` | `0.875` | `87.5%` |
| `{x:,}` | `1250000` | `1,250,000` |
| `{x:>8}` / `{x:<8}` | `"ab"` | alinear der./izq. |
| `{x!r}` | `"hola"` | `'hola'` (repr) |

🧠 **Para IA/datos:** la *normalización de texto* (minúsculas, strip, split) es
el primer paso de casi todo pipeline de NLP; encapsúlala en una función.

```python
texto = "  La IA necesita datos claros.  "
limpio = texto.strip()               # quita espacios de los extremos
palabras = limpio.lower().split()    # minúsculas + separa por espacios

print(repr(texto))
print(repr(limpio))
print(palabras)
print(f"{len(palabras)} palabras, {len(limpio)} caracteres")
```

```python
# Mini-lenguaje de formato en acción
precision = 0.875
grande = 1_250_000
print(f"porcentaje: {precision:.1%}")
print(f"2 decimales: {precision:.2f}")
print(f"miles: {grande:,}")
print(f"alineado: [{'ab':>8}] [{'ab':<8}]")
```

```python
# Métodos de cadena más usados
s = "Reseña: Excelente servicio"
print(s.lower())
print(s.replace("Excelente", "Pésimo"))
print(s.split(":"))
print(" | ".join(["a", "b", "c"]))
print("excelente" in s.lower())   # pertenencia de subcadena
```

## 4. Colecciones: list, tuple, dict, set

**Idea clave.** Elegir la colección correcta *es* una decisión de diseño.

| Tipo | Orden | Mutable | Duplicados | Uso típico | Acceso |
|---|---|---|---|---|---|
| `list` | Sí | Sí | Sí | secuencia que cambia | índice `l[0]` |
| `tuple` | Sí | No | Sí | registro fijo | índice |
| `dict` | Sí (inserción) | Sí | claves únicas | clave→valor | clave `d["k"]` |
| `set` | No | Sí | no | unicidad / pertenencia | `x in s` |

**Cómo decidir rápido:** ¿orden y cambio? → list · ¿registro fijo? → tuple ·
¿clave→valor? → dict · ¿solo pertenencia/unicidad? → set.

⚠️ `{}` es un **dict** vacío; el set vacío es `set()`.

🧠 **Para IA/datos:** el patrón rey para registros es **lista de diccionarios**
(`[{...}, {...}]`): conserva orden y campos con nombre. Es el paso previo a un
`DataFrame` de pandas.

```python
# LISTA: ordenada y mutable. Índices y cortes.
etiquetas = ["positivo", "neutral", "negativo", "positivo"]
print("primero:", etiquetas[0], "| último:", etiquetas[-1])
print("corte [1:3]:", etiquetas[1:3])   # incluye inicio, excluye fin
print("saltos [::2]:", etiquetas[::2])
etiquetas.append("mixto")
print("tras append:", etiquetas)
```

```python
# TUPLA (registro fijo) y DICT (campos nombrados)
punto = (20.97, -89.62)            # (lat, lon) de Mérida, Yucatán
lat, lon = punto                   # desempaquetado
print("lat:", lat, "lon:", lon)

prediccion = {"texto": "Excelente", "etiqueta": "positivo", "score": 0.97}
print("etiqueta:", prediccion["etiqueta"])
print("idioma (con default):", prediccion.get("idioma", "desconocido"))
print("¿tiene 'score'?:", "score" in prediccion)
```

```python
# SET: unicidad + operaciones de conjunto
permitidas = {"positivo", "neutral", "negativo"}
observadas = {"positivo", "positivo", "desconocido"}   # el duplicado se colapsa
print("observadas (sin duplicados):", observadas)
print("reconocidas (&):", observadas & permitidas)
print("no reconocidas (-):", observadas - permitidas)
print("todas (|):", observadas | permitidas)
```

## 5. Mutabilidad, alias y copias

**Idea clave.** Dos nombres pueden apuntar al **mismo** objeto (alias). Si es
mutable, modificarlo por un nombre lo cambia para todos.

| Necesidad | Herramienta |
|---|---|
| otro nombre del mismo objeto | `b = a` |
| contenedor nuevo, un nivel | `a.copy()`, `list(a)`, `dict(a)` |
| estructura anidada independiente | `copy.deepcopy(a)` |

🧠 **Para IA/datos:** causa #1 de bugs silenciosos en pipelines: transformas una
copia creyendo que el original está a salvo, y no lo estaba. Ante la duda,
`deepcopy`.

```python
original = ["a", "b"]
alias = original            # MISMO objeto
copia = original.copy()     # objeto NUEVO

alias.append("c")           # afecta a original
copia.append("z")           # NO afecta a original

print("original:", original)
print("alias:   ", alias)
print("copia:   ", copia)
print("original is alias:", original is alias)   # True
print("original is copia:", original is copia)   # False
```

```python
# ⚠️ Copia SUPERFICIAL: comparte los objetos internos
registro = {"texto": "hola", "etiquetas": ["saludo"]}
superficial = registro.copy()
superficial["etiquetas"].append("corto")
print("superficial afecta al original:", registro["etiquetas"])  # ['saludo','corto'] ⚠️

# ✅ Copia PROFUNDA: independiente en todos los niveles
import copy
registro = {"texto": "hola", "etiquetas": ["saludo"]}
profunda = copy.deepcopy(registro)
profunda["etiquetas"].append("corto")
print("deepcopy NO afecta al original:", registro["etiquetas"])  # ['saludo'] ✅
```

## 6. Decisiones y *truthiness* (cero ≠ ausencia)

**Idea clave.** Python trata como **falsos**: `False`, `None`, `0`, `0.0`,
`""`, `[]`, `{}`, `()`, `set()`. Todo lo demás es verdadero.

| Quiero preguntar… | Escribe | NO uses |
|---|---|---|
| ¿falta el valor? | `x is None` | `if not x` |
| ¿lista vacía? | `if not lista` | — |
| ¿el número es cero? | `x == 0` | `if not x` |

🧠 **Para IA/datos:** `score = 0.0` es un resultado **medido**; `score = None`
es *"no se evaluó"*. Confundirlos arruina promedios, filtros y métricas.

```python
# Cadena vacía es "falsa" -> útil para ramas de estado
for respuesta in ["", "ok", "una respuesta bastante larga de ejemplo"]:
    if not respuesta:
        estado = "sin respuesta"
    elif len(respuesta) < 20:
        estado = "respuesta breve"
    else:
        estado = "respuesta disponible"
    print(f"{respuesta!r:45} -> {estado}")
```

```python
# ⚠️ La trampa: 0.0 también es "falso"
score = 0.0
print("if not score  ->", "lo trata como ausente (MAL)" if not score else "ok")
print("score is None ->", "realmente ausente" if score is None else "hay un valor real:", score)
```

## 7. Ciclos y comprehensions

**Idea clave.** Recorre elementos directamente; usa `enumerate` para la posición
y `zip` para avanzar en paralelo. Las *comprehensions* expresan transformaciones
pequeñas y legibles.

| Usa comprehension si… | Usa ciclo `for` si… |
|---|---|
| transformación simple en una línea | hay varios pasos/decisiones |
| sin efectos secundarios | necesitas `print`/logging/`break` |

🔗 **Generadores:** cambia `[]` por `()` para producir valores de uno en uno sin
construir toda la lista (clave en datasets grandes).

```python
textos = ["muy bueno", "regular", "excelente"]
scores = [0.82, 0.51, 0.96]

print("-- enumerate (posición + valor) --")
for i, t in enumerate(textos, start=1):
    print(i, t)

print("-- zip (dos secuencias a la vez) --")
for t, s in zip(textos, scores):
    print(f"{t:<10} -> {s:.0%}")
```

```python
# Comprehensions: transformar y filtrar
altos = [s for s in scores if s >= 0.80]
fahrenheit = [c * 9/5 + 32 for c in [18, 25, 31]]
longitudes = {t: len(t) for t in textos}      # dict comprehension
iniciales = {t[0] for t in textos}            # set comprehension

print("altos:", altos)
print("fahrenheit:", fahrenheit)
print("longitudes:", longitudes)
print("iniciales:", iniciales)
```

```python
# 🔗 Generador: no crea lista intermedia en memoria
total = sum(s for s in scores if s >= 0.8)
print("suma de scores >= 0.8:", round(total, 2))
gen = (s * 2 for s in scores)
print("tipo:", type(gen).__name__, "| primer valor:", next(gen))
```

## 8. Funciones

**Idea clave.** Una función nombra una transformación: recibe parámetros,
devuelve una salida con `return` (sin `return` devuelve `None`). Reduce
repetición y permite probar ideas aisladas.

El `*` en la firma fuerza a pasar los argumentos siguientes **por nombre**, lo
que hace el código del llamador autoexplicativo.

⚠️ **Trampa clásica:** nunca uses un valor por defecto **mutable** (`def f(x=[])`):
se comparte entre llamadas. Usa `None` como centinela.

🧠 **Para IA/datos:** una función *pura* (misma entrada → misma salida, sin
efectos secundarios) es fácil de probar y reutilizar en un pipeline.

```python
def normalizar_texto(texto, *, minusculas=True):
    """Quita espacios externos y, opcionalmente, pasa a minúsculas."""
    resultado = texto.strip()
    if minusculas:
        resultado = resultado.lower()
    return resultado

print(normalizar_texto("  Hola IA  "))
print(normalizar_texto("  Hola IA  ", minusculas=False))

# assert: verifica tus propias expectativas (no entradas de usuario)
assert normalizar_texto("  HOLA  ") == "hola"
assert normalizar_texto("") == ""
print("asserts OK ✅")
```

```python
# ⚠️ Default mutable: la MISMA lista persiste entre llamadas
def agregar_mal(x, bolsa=[]):
    bolsa.append(x)
    return bolsa
print("agregar_mal(1):", agregar_mal(1))
print("agregar_mal(2):", agregar_mal(2), "  <- ¡arrastró el 1!")

# ✅ Centinela None + lista nueva adentro
def agregar_bien(x, bolsa=None):
    if bolsa is None:
        bolsa = []
    bolsa.append(x)
    return bolsa
print("agregar_bien(1):", agregar_bien(1))
print("agregar_bien(2):", agregar_bien(2), "  <- independiente ✅")
```

## 9. Ordenar por criterio

**Idea clave.** `sorted(iterable, key=...)` devuelve una lista **nueva**;
`lista.sort()` ordena **en el lugar**. `key` recibe una función (se pasa *sin
llamarla*) que calcula el criterio de cada elemento.

| Quiero… | Usa |
|---|---|
| conservar el original y obtener uno nuevo | `sorted(x)` |
| reordenar el que ya tengo (ahorra memoria) | `x.sort()` |
| ordenar algo que no es lista | `sorted(x)` |

🧠 El orden de Python es *estable*: los empates conservan su orden previo.

```python
evaluaciones = [
    {"modelo": "a", "score": 0.70},
    {"modelo": "b", "score": 0.95},
    {"modelo": "c", "score": 0.80},
]

# lambda = función anónima pequeña, en el mismo lugar
ranking = sorted(evaluaciones, key=lambda e: e["score"], reverse=True)
for e in ranking:
    print(f"{e['modelo']} -> {e['score']}")
```

```python
palabras = ["pera", "uva", "manzana"]
print("por longitud:", sorted(palabras, key=len))
print("alfabético:  ", sorted(palabras, key=str.lower))

# Varias claves con una tupla: por score desc y, a igualdad, por modelo
datos = [{"m": "a", "s": 0.9}, {"m": "b", "s": 0.9}, {"m": "c", "s": 0.8}]
print("multi-criterio:", [d["m"] for d in sorted(datos, key=lambda d: (-d["s"], d["m"]))])
```

## 10. Módulos, imports y la librería estándar

**Idea clave.** La librería estándar ("baterías incluidas") trae muchísimo;
explórala antes de instalar dependencias. Importa el módulo o nombres
concretos; evita `from modulo import *` (oculta el origen de los nombres).

| Módulo | Para qué |
|---|---|
| `math` | raíz, log, `isclose`, constantes |
| `statistics` | media, mediana, desviación |
| `collections` | `Counter`, `defaultdict`, `deque` |
| `itertools` | combinaciones, productos, agrupaciones |
| `pathlib` | rutas como objetos (`Path`) |
| `json` | leer/escribir JSON |
| `random` | muestreo, barajar, semillas |

🧠 **Para IA/datos:** `Counter` y `defaultdict(list)` resuelven el 80% de los
conteos y agrupaciones antes de necesitar pandas.

```python
from collections import Counter, defaultdict
from statistics import mean

etiquetas = ["positivo", "neutral", "negativo", "positivo"]

# Counter: frecuencias
c = Counter(etiquetas)
print("frecuencias:", c)
print("más común:", c.most_common(1))

# defaultdict(list): agrupar sin inicializar claves a mano
registros = [
    {"etiqueta": "positivo", "score": 0.9},
    {"etiqueta": "neutral",  "score": 0.6},
    {"etiqueta": "positivo", "score": 0.8},
]
por_etiqueta = defaultdict(list)
for r in registros:
    por_etiqueta[r["etiqueta"]].append(r["score"])
print("agrupado:", dict(por_etiqueta))

# De agrupar a resumir
promedios = {etq: round(mean(ss), 3) for etq, ss in por_etiqueta.items()}
print("promedios:", promedios)
```

## 11. Excepciones y tracebacks

**Idea clave.** Ante un error, Python lanza una excepción y muestra un
**traceback**. Léelo **de abajo hacia arriba**: la última línea da el tipo y
el mensaje.

| Excepción | Causa típica |
|---|---|
| `ValueError` | valor inválido: `int("tres")` |
| `TypeError` | tipos incompatibles: `"x" + 1` |
| `KeyError` | clave inexistente: `d["falta"]` |
| `IndexError` | índice fuera de rango: `lista[99]` |
| `FileNotFoundError` | archivo que no existe |

**Reglas de oro:** captura la excepción **específica**; nunca `except: pass`
(convierte un fallo visible en datos malos); conserva la causa con
`raise ... from error`.

🧠 **Para IA/datos:** en limpieza tendrás filas corruptas. Decide
explícitamente: ¿se salta, se marca o se detiene el proceso?

```python
# ✅ Captura concreta + conserva la causa
def parse_cantidad(texto):
    try:
        return int(texto)
    except ValueError as error:
        raise ValueError(f"Cantidad inválida: {texto!r}") from error

print("parse_cantidad('3') =", parse_cantidad("3"))
try:
    parse_cantidad("tres")
except ValueError as error:
    print("error esperado:", error)
```

```python
# ⚠️ Antipatrón: esconder TODO -> un valor inválido se vuelve un 0 "real"
def parse_mal(texto):
    try:
        return int(texto)
    except Exception:
        return 0
print("parse_mal('tres') =", parse_mal("tres"), " <- dato corrupto silencioso")

# ✅ Silenciar SOLO lo esperado (config opcional ausente -> default)
from pathlib import Path
from tempfile import TemporaryDirectory
def leer_config(path):
    try:
        return path.read_text(encoding="utf-8")
    except FileNotFoundError:
        return "theme=light"   # ausente = usar default documentado
with TemporaryDirectory() as d:
    cfg = Path(d) / "config.txt"
    print("sin archivo:", leer_config(cfg))
    cfg.write_text("theme=dark", encoding="utf-8")
    print("con archivo:", leer_config(cfg))
```

## 12. Programación orientada a objetos (POO)

**Idea clave.** Una **clase** define cómo construir objetos y qué saben hacer.
Cada **instancia** es un objeto concreto; sus **atributos** guardan datos y sus
**métodos** son funciones asociadas. `__init__` inicializa la instancia y `self`
es esa instancia dentro de los métodos.

**Cuándo usar una clase (y cuándo no):** usa una clase cuando tienes *estado*
que cambia con el tiempo y *comportamiento* que opera sobre él (un contador, un
registro que acumula). Para una transformación sin estado, una **función** basta.

⚠️ **Trampa:** un atributo definido en el cuerpo de la clase se **comparte**
entre instancias. Los datos propios deben crearse en `__init__`.

```python
class Contador:
    def __init__(self, nombre):
        self.nombre = nombre      # atributo propio de cada instancia
        self.total = 0

    def agregar(self):
        self.total += 1

    def reiniciar(self):
        self.total = 0

a = Contador("lote A")
b = Contador("lote B")
a.agregar(); a.agregar()
print(a.nombre, a.total)   # lote A 2
print(b.nombre, b.total)   # lote B 0  (independiente)
```

```python
# Alias de una instancia: NO crea un objeto nuevo
a = Contador("lote A"); a.agregar(); a.agregar()
alias = a
alias.reiniciar()
print("a.total:", a.total, "| alias is a:", alias is a)   # 0 | True
```

```python
# ⚠️ Lista compartida por accidente (atributo de CLASE)
class RegistroCompartido:
    registros = []            # pertenece a la CLASE
    def agregar(self, x): self.registros.append(x)

ca, cb = RegistroCompartido(), RegistroCompartido()
ca.agregar("dato de A")
print("compartido -> cb ve:", cb.registros, "| mismo objeto:", ca.registros is cb.registros)

# ✅ Estado independiente: inicializa la lista en __init__
class RegistroIndependiente:
    def __init__(self): self.registros = []
    def agregar(self, x): self.registros.append(x)

ia, ib = RegistroIndependiente(), RegistroIndependiente()
ia.agregar("dato de A")
print("independiente -> ib ve:", ib.registros, "| mismo objeto:", ia.registros is ib.registros)
```

## 13. Caso integrador: registro de predicciones

Junta todo: una clase con **estado** (`_registros`), que al agregar guarda una
**copia** con su posición, y un método `resumen()` que combina `Counter`,
promedio y el manejo correcto de `None` (promedio `None` si no hay registros).

🧠 El guion bajo en `_registros` es una convención: "esto es interno, no lo
toques desde fuera". Python no lo bloquea, pero comunica la intención.

```python
from collections import Counter

class RegistroPredicciones:
    def __init__(self, nombre):
        self.nombre = nombre
        self._registros = []

    def agregar(self, registro, posicion):
        guardado = registro.copy()        # no mutar el dict original del llamador
        guardado["posicion"] = posicion
        self._registros.append(guardado)

    def resumen(self):
        scores = [r["score"] for r in self._registros]
        return {
            "cantidad": len(self._registros),
            "conteos": dict(Counter(r["etiqueta"] for r in self._registros)),
            "promedio": sum(scores) / len(scores) if scores else None,
        }

predicciones = [
    {"texto": "Excelente",   "etiqueta": "positivo", "score": 0.98},
    {"texto": "Regular",     "etiqueta": "neutral",  "score": 0.66},
    {"texto": "Llegó roto",  "etiqueta": "negativo", "score": 0.93},
    {"texto": "Cero válido", "etiqueta": "neutral",  "score": 0.0},
]

lote = RegistroPredicciones("Demostración")
for pos, p in enumerate(predicciones):
    lote.agregar(p, pos)

print(lote.resumen())
```

```python
# Comprobaciones del contrato (como en el ejercicio 11)
r = lote.resumen()
assert r["cantidad"] == 4
assert r["conteos"] == {"positivo": 1, "neutral": 2, "negativo": 1}
assert abs(r["promedio"] - 0.6425) < 1e-9
assert 0.0 in [x["score"] for x in lote._registros]   # el 0.0 se conserva

vacio = RegistroPredicciones("vacío")
assert vacio.resumen()["promedio"] is None             # sin datos -> None
print("contrato verificado ✅  | lote vacío:", vacio.resumen())
```

## 14. El Zen de Python como criterio de diseño

El Zen (`import this`, PEP 20) no es un reglamento sino **criterios** para
decidir entre dos soluciones que funcionan. Las 19 reglas, resumidas:

| # | Regla | En una frase |
|---|---|---|
| 1 | Bonito > feo | el formato claro revela la intención |
| 2 | Explícito > implícito | haz visibles dependencias y decisiones |
| 3 | Simple > complejo | si hay built-in que lo hace, úsalo |
| 4 | Complejo > complicado | organiza los pasos, no los comprimas |
| 5 | Plano > anidado | *guard clauses* reducen indentación |
| 6 | Disperso > denso | espacios y saltos ayudan a leer |
| 7 | La legibilidad cuenta | nombres que describen el dominio |
| 8 | Los casos especiales no rompen reglas | mantén contratos consistentes |
| 9 | …pero la practicidad vence a la pureza | adapta la salida donde ayuda |
| 10 | Los errores no pasan en silencio | no escondas excepciones |
| 11 | …salvo silenciarlos explícitamente | silencia solo lo esperado |
| 12 | Ante la ambigüedad, no adivines | exige un contrato de entrada |
| 13 | Una manera obvia de hacerlo | usa la convención del lenguaje |
| 14 | …aunque no sea obvia al principio | lo "obvio" requiere práctica |
| 15 | Ahora > nunca | empieza con algo ejecutable |
| 16 | …aunque nunca > *ahora mismo* | no corras sin medir el costo |
| 17 | Si es difícil de explicar, es mala idea | rediseña o divide |
| 18 | Si es fácil de explicar, *puede* ser buena | la claridad no prueba corrección |
| 19 | Los espacios de nombres son geniales | `math.sqrt` dice de dónde viene |

Abajo, las reglas más prácticas con su **antes/después** ejecutable.

```python
# Regla 5 (plano > anidado): guard clauses con retorno temprano
def puede_editar_anidado(user, doc):
    if user["activo"]:
        if not doc["archivado"]:
            if doc["dueno"] == user["id"]:
                return True
            else: return False
        else: return False
    else: return False

def puede_editar(user, doc):                 # ✅ mismo resultado, plano
    if not user["activo"]:   return False
    if doc["archivado"]:     return False
    return doc["dueno"] == user["id"]

u = {"id": 7, "activo": True}; d = {"dueno": 7, "archivado": False}
print("plano:", puede_editar(u, d), "| anidado:", puede_editar_anidado(u, d))
```

```python
# Regla 7 (legibilidad): nombres con significado y unidades
pos_inicial_m, vel_m_s, t_s = 10, 3, 4
pos_final_m = pos_inicial_m + vel_m_s * t_s
print("legible  ->", pos_final_m, "m")

p, v, t = 10, 3, 4            # ⚠️ ¿qué son? hay que adivinar
print("críptico ->", p + v * t)
```

```python
# Regla 8 (contratos consistentes): devuelve SIEMPRE el mismo tipo
def activos_inconsistente(users):
    nombres = [u["n"] for u in users if u["activo"]]
    return nombres if nombres else "sin activos"   # ⚠️ a veces list, a veces str

def activos(users):                                 # ✅ siempre lista
    return [u["n"] for u in users if u["activo"]]

print("inconsistente([]):", repr(activos_inconsistente([])))
print("consistente([]):  ", activos([]))
# La capa de presentación decide el mensaje, sin romper el contrato:
nombres = activos([])
print("mostrar:", ", ".join(nombres) if nombres else "sin activos")
```

```python
# Regla 10 (errores visibles) + 5 + 7 juntas: refactor del ejercicio del Zen
def valid_scores(records):
    if records is None:
        return None
    scores = []
    for record in records:
        if record is None or "score" not in record:
            continue
        try:
            score = float(record["score"])
        except (ValueError, TypeError):
            continue                       # captura SOLO lo que sabemos manejar
        if score < 0:
            continue
        scores.append(score)
    return sum(scores) / len(scores) if scores else None

datos = [{"score": "0.9"}, None, {}, {"score": "invalid"}, {"score": 0.7}, {"score": -0.5}]
print("promedio válido:", valid_scores(datos))   # 0.8
print("sin válidos:", valid_scores([]), "| None:", valid_scores(None))
```

### 15.1 Anotaciones de tipo (type hints)

Documentan la *intención* sin cambiar el comportamiento (Python no las obliga en
tiempo de ejecución; herramientas como `mypy` las revisan). Hacen el código
autoexplicativo y habilitan el autocompletado del editor.

```python
def promedio_o_none(valores: list[float]) -> float | None:
    """Devuelve el promedio, o None si la lista está vacía."""
    if not valores:
        return None
    return sum(valores) / len(valores)

print(promedio_o_none([0.5, 1.0]))   # 0.75
print(promedio_o_none([]))           # None
# Nota: las anotaciones NO validan en runtime; son documentación + herramientas.
```

### 15.2 `dataclass`: registros con menos ceremonia

Cuando un objeto es básicamente "datos con nombre", `@dataclass` genera
`__init__`, `__repr__` y comparación por ti. Ideal para registros de un pipeline.

```python
from dataclasses import dataclass

@dataclass
class Prediccion:
    texto: str
    etiqueta: str
    score: float

p = Prediccion("Excelente", "positivo", 0.97)
print(p)                 # repr automático y legible
print(p.score, "->", p.etiqueta)
print(p == Prediccion("Excelente", "positivo", 0.97))   # comparación por valor
```

### 15.3 Context managers (`with`)

Garantizan que un recurso se **cierre** aunque haya un error (archivos,
conexiones). Ya lo viste con `TemporaryDirectory`; el patrón es el mismo con
archivos.

```python
from pathlib import Path
from tempfile import TemporaryDirectory

with TemporaryDirectory() as d:
    ruta = Path(d) / "notas.txt"
    with ruta.open("w", encoding="utf-8") as f:   # se cierra solo al salir del with
        f.write("linea 1\nlinea 2\n")
    print("contenido leído:", ruta.read_text(encoding="utf-8").splitlines())
```

## 16. Chuleta final

**Elegir estructura de datos**

- Secuencia que cambia → `list` · Registro fijo → `tuple`
- Clave→valor → `dict` · Unicidad / pertenencia → `set`

**Cero vs ausencia**

- Valor medido `0` / `0.0` → número real · *No medido* → `None`
- ¿Falta? `x is None` (no `if not x`)

**Copias**

- Mismo objeto → `b = a` · Un nivel → `a.copy()` · Anidado → `copy.deepcopy(a)`

**Funciones**

- Nunca default mutable (`def f(x=[])`) → usa `None` como centinela
- `*` en la firma → argumentos por nombre, más legible

**Errores**

- Captura la excepción **específica**; nunca `except: pass`
- `raise ... from error` conserva la causa

**POO**

- Datos propios en `__init__` (no en el cuerpo de la clase)
- ¿Estado + comportamiento? clase · ¿solo transformar? función

**Zen, en tres ideas**

- Explícito > implícito · Plano > anidado · Los errores no pasan en silencio
